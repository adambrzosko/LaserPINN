# 10 GHz Phase-Randomised GS-DFB — Interactive Re-plotting

Loads 1M-pulse frequency sweep data from `images/acceptance_bw_comparison/data/`
and power sweep / spectra / detection chain from `images/paper_10ghz/data/`.

Each figure section has **tweakable parameters** in a clearly marked cell. Edit those and re-run.

```python
import numpy as np
import matplotlib.pyplot as plt
import json
from pathlib import Path

# ── 1M-pulse frequency sweep (acceptance BW comparison) ───────
ABW_DATA = Path('../images/acceptance_bw_comparison_v2/data')
ACCEPTANCE_BW = 0.6   # nm — change to 0.6 or 3.0 to switch

free_1m = np.load(ABW_DATA / 'free_running.npz', allow_pickle=True)
sld_1m  = np.load(ABW_DATA / f'sld_bw{ACCEPTANCE_BW:.1f}nm.npz', allow_pickle=True)
scalars = json.loads((ABW_DATA / f'scalars_bw{ACCEPTANCE_BW:.1f}nm.json').read_text())

freq_labels = list(free_1m['freq_labels'])
N_PULSES = int(free_1m['N_PULSES'])

# Build unified dict matching the key pattern used by figures:
#   '{flabel}_{free|sld}_eta', '{flabel}_{free|sld}_autocorr', etc.
freq_sweep = {}
for flabel in freq_labels:
    for suffix in ['eta', 'dphi', 'autocorr', 'eta_autocorr']:
        freq_sweep[f'{flabel}_free_{suffix}'] = free_1m[f'{flabel}_{suffix}']
        freq_sweep[f'{flabel}_sld_{suffix}']  = sld_1m[f'{flabel}_{suffix}']

print(f'Frequencies: {freq_labels}')
print(f'N_PULSES = {N_PULSES:,}  (acceptance BW = {ACCEPTANCE_BW} nm)')

# ── Paper 10 GHz data (power sweep, spectra, detection chain) ─
DATA = Path('../images/paper_10ghz/data')
OUT  = Path('../images/paper_10ghz')

params       = json.loads((DATA / 'parameters.json').read_text())
power_sweep  = np.load(DATA / 'power_sweep.npz')
mechanism    = np.load(DATA / 'mechanism.npz')
det_1g       = np.load(DATA / 'detection_1_GHz_free.npz')
det_10g      = np.load(DATA / 'detection_10_GHz_plus_SLD.npz')
det_enob     = np.load(DATA / 'detection_enob_sweep.npz')

spec_data = {}
for p in DATA.glob('spectrum_*.npz'):
    key = p.stem.replace('spectrum_', '').replace('_', ' ')
    spec_data[key] = np.load(p)

print(f'Spectrum configs: {list(spec_data.keys())}')
```

```python
# Helpers

def eta_autocorrelation(eta, max_lag=50):
    """Compute normalised autocovariance of eta."""
    c = eta - np.mean(eta)
    v = np.var(eta)
    if v < 1e-30:
        return np.zeros(max_lag)
    ac = np.empty(max_lag)
    ac[0] = 1.0
    for k in range(1, max_lag):
        ac[k] = np.mean(c[k:] * c[:-k]) / v
    return ac

def autocorr_ci(n, confidence=0.99):
    """Rayleigh-based CI for |r_k| under H0: uniform iid phases."""
    return np.sqrt(-np.log(1.0 - confidence) / n)

def bartlett_ci(n, confidence=0.99):
    """Bartlett CI for eta autocorrelation: ±z_{alpha/2}/sqrt(N)."""
    from scipy.stats import norm
    z = norm.ppf(1 - (1 - confidence) / 2)
    return z / np.sqrt(n)

# Theoretical arcsine PDF
eta_th = np.linspace(0.001, 0.999, 500)
arcsine_pdf = 1.0 / (np.pi * np.sqrt(eta_th * (1.0 - eta_th)))
```

```python
plt.rcParams.update({
    'font.size': 14,
    'axes.labelsize': 14,
    'axes.titlesize': 14,
    'xtick.labelsize': 14,
    'ytick.labelsize': 14,
    'xtick.direction': 'in',
    'ytick.direction': 'in',
    'xtick.major.size': 6,
    'ytick.major.size': 6,
    'xtick.major.width': 1.2,
    'ytick.major.width': 1.2,
    'xtick.minor.size': 3,
    'ytick.minor.size': 3,
    'xtick.minor.width': 1.0,
    'ytick.minor.width': 1.0,
    'xtick.minor.visible': True,
    'ytick.minor.visible': True,
    'axes.grid': True,
    'legend.fontsize': 14,
})
```

## Figure 1 — AMZI Histograms + η Autocorrelation

```python
# ─── Tweakable parameters ───────────────────────────────────────
FIG1_BINS       = 256
FIG1_YLIM_HIST  = 12e-3
FIG1_MAX_LAG    = 100
FIG1_CI_LEVEL   = 0.99       # confidence level for eta autocorrelation
FIG1_FIGSIZE    = (4, 3.5)   # per-panel size (width, height)
FIG1_HIST_ALPHA = 0.7
FIG1_DPI        = 150
FIG1_SAVE       = False
# ────────────────────────────────────────────────────────────────
```

```python
freq_labels = freq_labels[0::2]
```

```python
freq_labels
```

```python
from matplotlib.ticker import ScalarFormatter
fmt = ScalarFormatter(useMathText=True)
fmt.set_scientific(True)
fmt.set_powerlimits((0, 0))

n_freq = 3 # en(freq_labels)
ci99_eta = bartlett_ci(N_PULSES, FIG1_CI_LEVEL)

fig1, axes1 = plt.subplots(
    4, n_freq,
    figsize=(FIG1_FIGSIZE[0] * n_freq, FIG1_FIGSIZE[1] * 4))

for col, flabel in enumerate(freq_labels):
    for blk, inj_label in enumerate(['free', 'sld']):
        eta = freq_sweep[f'{flabel}_{inj_label}_eta']
        r1 = scalars[f'{flabel}_{inj_label}']['r1']
        eta_ac = eta_autocorrelation(eta, FIG1_MAX_LAG)

        row_hist = 2 * blk
        row_ac   = 2 * blk + 1

        # Histogram
        ax_h = axes1[row_hist, col]
        color = 'C0' if blk == 0 else 'C1'
        ax_h.hist(eta, bins=FIG1_BINS, weights=np.ones_like(eta)/len(eta),
                  alpha=FIG1_HIST_ALPHA, color=color,
                  label=f'Sim (r$_1$={r1:.3f})')
        bin_width = 1.0 / FIG1_BINS
        arcsine_scaled = arcsine_pdf * bin_width
        ax_h.plot(eta_th, arcsine_scaled, 'k--', lw=1.2, alpha=0.6,
                  label='Ideal arcsine')
        ax_h.set_xlim(0, 1)
        ax_h.set_ylim(0, FIG1_YLIM_HIST)
        if col == 0:
            ax_h.set_ylabel('Probability density')
        ax_h.legend(fontsize=14, loc='upper center')
        if row_hist < 2:
            ax_h.set_xticklabels([])
        else:
            ax_h.set_xlabel('Intensity')
        
        # Autocorrelation
        ax_a = axes1[row_ac, col]
        lags = np.arange(len(eta_ac))
        ax_a.plot(lags[1:], eta_ac[1:], 'o-', ms=3, lw=1,
          color='C0' if blk == 0 else 'C1', alpha=0.7)
        ax_a.axhline(+ci99_eta, color='g', ls='--', lw=1)
        ax_a.axhline(-ci99_eta, color='g', ls='--', lw=1,
                     label=f'{FIG1_CI_LEVEL*100:.0f}% CI')
        ax_a.set_xlim(0, FIG1_MAX_LAG)
        if col == 0:
            ax_a.set_ylabel('Autocorrelation')
        ymax = max(0.01, 1.3 * np.max(np.abs(eta_ac[1:])))
        ax_a.set_ylim(-min(ymax, 1.0), min(ymax, 1.0))
        if row_ac < 3:
            ax_a.set_xticklabels([])
        else:
            ax_a.set_xlabel('Lag')
        ax_a.legend(fontsize=14, loc='upper right')
        
plt.tight_layout()
if FIG1_SAVE:
    fig1.savefig(OUT / 'sim_phase_randomisation.png',
                 dpi=FIG1_DPI, bbox_inches='tight')
    print(f'Saved: {OUT}/sim_phase_randomisation.png')
plt.show()
```

## Figure 2 — Phase Autocorrelation (Selected Cases)

```python
# ─── Tweakable parameters ───────────────────────────────────────
FIG2_CASES = [
    ('1 GHz_free',  '1 GHz — Free (reference)'),
    ('10 GHz_free', '10 GHz — Free'),
    ('10 GHz_sld',  '10 GHz — SLD injected'),
]
FIG2_CI_LEVELS   = [0.99, 0.999]
FIG2_MAX_LAG     = 50
FIG2_FIGSIZE     = (18, 5)
FIG2_DPI         = 150
FIG2_SAVE        = False
# ────────────────────────────────────────────────────────────────
```

```python
fig2, axes2 = plt.subplots(1, len(FIG2_CASES), figsize=FIG2_FIGSIZE)

ci_vals = {c: autocorr_ci(N_PULSES, c) for c in FIG2_CI_LEVELS}

for ax, (key, title) in zip(axes2, FIG2_CASES):
    ac = freq_sweep[f'{key}_autocorr']
    lags = np.arange(min(len(ac), FIG2_MAX_LAG))
    color = 'C0' if 'free' in key else 'C1'

    ax.bar(lags[1:], ac[1:len(lags)], width=0.8,
           color=color, alpha=0.7)

    styles = ['--', ':']
    for (conf, ci_val), ls in zip(ci_vals.items(), styles):
        ax.axhline(ci_val, color='r', ls=ls, lw=1,
                   label=f'{conf*100:.1f}% CI = {ci_val:.4f}')

    ax.set_xlabel('Lag $k$')
    ax.set_ylabel('$|r_k|$')
    ax.legend(fontsize=14)
    ax.set_xlim(0, FIG2_MAX_LAG)
    ymax = max(0.01, 1.5 * np.max(ac[1:len(lags)]))
    ax.set_ylim(0, min(ymax, 1.0))

plt.tight_layout()
if FIG2_SAVE:
    fig2.savefig(OUT / 'autocorrelation.png',
                 dpi=FIG2_DPI, bbox_inches='tight')
    print(f'Saved: {OUT}/autocorrelation.png')
plt.show()
```

## Figure 3 — Injection Power Sweep at 10 GHz

```python
# ─── Tweakable parameters ───────────────────────────────────────
FIG3_FIGSIZE      = (18, 10)
FIG3_MARKER_SIZE  = 6
FIG3_LINE_WIDTH   = 2
FIG3_DPI          = 150
FIG3_SAVE         = False

# Experimental jitter values from the paper (for overlay)
PAPER_P_MW     = [0, 5, 19, 24]
PAPER_JITTER   = [4.4, 15.1, 21.9, 28.3]
# ────────────────────────────────────────────────────────────────
```

```python
sw = power_sweep
N_sw = params['simulation']['N_pulses_power_sweep']
ci99_sw = autocorr_ci(N_sw, 0.99)

fig3, axes3 = plt.subplots(2, 3, figsize=FIG3_FIGSIZE)

ms = FIG3_MARKER_SIZE
lw = FIG3_LINE_WIDTH

# (a) r1 vs power
ax = axes3[0, 0]
ax.plot(sw['P_mW'], sw['r1'], 'o-', color='C0', lw=lw, ms=ms)
ax.axhline(ci99_sw, color='r', ls='--', lw=1, alpha=0.7,
           label=f'99% CI = {ci99_sw:.4f}')
ax.set_ylabel('Phase correlation $r_1$')
ax.legend(fontsize=14)
ax.set_ylim(bottom=0)

# (b) KL divergence
ax = axes3[0, 1]
ax.plot(sw['P_mW'], sw['kl'], 's-', color='C2', lw=lw, ms=ms)
ax.set_ylabel('KL divergence from uniform (bits)')
ax.set_ylim(bottom=0)

# (c) KS statistic
ax = axes3[0, 2]
ax.plot(sw['P_mW'], sw['ks_stat'], 'd-', color='C4', lw=lw, ms=ms)
ax.axhline(1.63 / np.sqrt(min(N_sw, 10000)),
           color='r', ls='--', lw=1, alpha=0.7, label='KS 1% threshold')
ax.set_ylabel('KS statistic vs uniform')
ax.legend(fontsize=14)

# (d) Timing jitter
ax = axes3[1, 0]
ax.plot(sw['P_mW'], sw['sigma_t_ps'], '^-', color='C1', lw=lw, ms=ms)
ax.set_ylabel('RMS timing jitter (ps)')

# (e) Intensity CV
ax = axes3[1, 1]
ax.plot(sw['P_mW'], sw['cv_P'], 'v-', color='C3', lw=lw, ms=ms)
ax.set_ylabel('Intensity CV')

# (f) Mean power
ax = axes3[1, 2]
ax.plot(sw['P_mW'], sw['mean_P_mW'], 'p-', color='C5', lw=lw, ms=ms)
ax.set_ylabel('Mean peak power (mW)')

for ax in axes3.flat:
    ax.set_xlabel('SLD injection power (mW)')

plt.tight_layout()
if FIG3_SAVE:
    fig3.savefig(OUT / 'injection_power_sweep.png',
                 dpi=FIG3_DPI, bbox_inches='tight')
    print(f'Saved: {OUT}/injection_power_sweep.png')
plt.show()
```

## Figure 4 — Optical Spectra

```python
# ─── Tweakable parameters ───────────────────────────────────────
FIG4_FREQ_RANGE_GHZ = 60    # show ± this around carrier
FIG4_YLIM_DB        = (-60, 5)
FIG4_FIGSIZE        = (18, 5)
FIG4_DPI            = 150
FIG4_SAVE           = False

FIG4_CONFIGS = [
    ('1 GHz free',  '1 GHz — Free (reference)', 'C0'),
    ('10 GHz free', '10 GHz — Free (comb lines)', 'C3'),
    ('10 GHz sld',  '10 GHz — SLD injected', 'C1'),
]
# ────────────────────────────────────────────────────────────────
```

```python
filter_bw_ghz = params['filter']['bw_ghz']
filter_bw_nm  = params['filter']['bw_nm']

fig4, axes4 = plt.subplots(1, len(FIG4_CONFIGS), figsize=FIG4_FIGSIZE)

for ax, (key, title, color) in zip(axes4, FIG4_CONFIGS):
    sd = spec_data[key]
    f_GHz = sd['freq_hz'] * 1e-9
    mask = np.abs(f_GHz) < FIG4_FREQ_RANGE_GHZ

    # Unfiltered (faint)
    spec_dB_raw = 10 * np.log10(np.maximum(sd['spectrum_raw'][mask], 1e-10))
    ax.plot(f_GHz[mask], spec_dB_raw, color=color, lw=0.3, alpha=0.25,
            label='Unfiltered')

    # Filtered (bold)
    spec_dB_filt = 10 * np.log10(
        np.maximum(sd['spectrum_filtered'][mask], 1e-10))
    ax.plot(f_GHz[mask], spec_dB_filt, color=color, lw=0.8, alpha=0.9,
            label=f'{filter_bw_nm} nm filtered')

    # Comb line markers
    f_rep_GHz = float(sd['f_rep']) * 1e-9
    for n in range(-6, 7):
        if n != 0:
            ax.axvline(n * f_rep_GHz, color='gray', ls=':', lw=0.5, alpha=0.3)

    # Filter edges
    ax.axvline(-filter_bw_ghz / 2, color='k', ls='--', lw=0.7, alpha=0.4)
    ax.axvline(+filter_bw_ghz / 2, color='k', ls='--', lw=0.7, alpha=0.4)

    ax.set_xlabel('Frequency offset (GHz)')
    ax.set_ylabel('Power spectral density (dB)')
    ax.set_ylim(*FIG4_YLIM_DB)
    ax.legend(fontsize=14, loc='upper right')

plt.tight_layout()
if FIG4_SAVE:
    fig4.savefig(OUT / 'optical_spectra.png',
                 dpi=FIG4_DPI, bbox_inches='tight')
    print(f'Saved: {OUT}/optical_spectra.png')
plt.show()
```

## Figure 5 — Mechanism Decomposition

```python
# ─── Tweakable parameters ───────────────────────────────────────
FIG5_FIGSIZE    = (18, 5)
FIG5_DPI        = 150
FIG5_SAVE       = True
FIG5_CMAP       = 'viridis'

# Experimental jitter values from paper (for overlay on panel b)
FIG5_PAPER_P       = [0, 5, 19, 24]
FIG5_PAPER_JITTER  = [4.4, 15.1, 21.9, 28.3]
# ────────────────────────────────────────────────────────────────
```

```python
m = mechanism
P_fine = m['P_fine_mW']
r1_fine = m['r1_fine']
jitter_fine = m['jitter_fine_ps']
P_threshold = float(m['P_threshold_mW'])
ci99_mech = float(m['ci99'])

fig5, axes5 = plt.subplots(1, 3, figsize=FIG5_FIGSIZE)

# (a) r1 vs SLD power
ax = axes5[0]
ax.plot(P_fine, r1_fine, 'o-', color='C0', ms=4, lw=1.5, label='Simulation')
ax.axhline(ci99_mech, color='r', ls='--', lw=1,
           label=f'99% CI = {ci99_mech:.4f}')
ax.axvspan(P_threshold, P_fine[-1], alpha=0.1, color='green',
           label='Phase-randomised regime')
ax.axvline(P_threshold, color='green', ls='--', lw=1)
ax.set_xlabel('SLD injection power (mW)')
ax.set_ylabel('$r_1$')
ax.legend(fontsize=14)
ax.set_ylim(bottom=0)

# (b) jitter vs SLD power
ax = axes5[1]
ax.plot(P_fine, jitter_fine, 's-', color='C0', ms=4, lw=1.5, label='Simulation')
ax.axvline(P_threshold, color='black', ls='--', lw=1,
           label=f'Threshold = {P_threshold:.0f} mW')
ax.scatter(FIG5_PAPER_P, FIG5_PAPER_JITTER, marker='*', s=150, c='red',
           zorder=5, label='Experiment')
ax.set_xlabel('SLD injection power (mW)')
ax.set_ylabel('RMS timing jitter (ps)')
ax.legend(fontsize=14)

# (c) r1 vs jitter (parametric)
ax = axes5[2]
sc = ax.scatter(jitter_fine, r1_fine, c=P_fine, cmap=FIG5_CMAP,
                s=40, edgecolors='k', lw=0.5, zorder=3)
ax.axhline(ci99_mech, color='r', ls='--', lw=1)
ax.set_xlabel('RMS timing jitter (ps)')
ax.set_ylabel('$r_1$')
cbar = plt.colorbar(sc, ax=ax)
cbar.set_label('SLD power (mW)')
ax.set_ylim(bottom=0)

plt.tight_layout()
if FIG5_SAVE:
    fig5.savefig(OUT / 'mechanism_decomposition.png',
                 dpi=FIG5_DPI, bbox_inches='tight')
    print(f'Saved: {OUT}/mechanism_decomposition.png')
plt.show()
```

```python
det_p = params['detection']

fig6, axes6 = plt.subplots(2, 2, figsize=FIG6_FIGSIZE)

# (a) η distribution — 1 GHz free
ax = axes6[0, 0]
ax.hist(det_1g['eta_true'], bins=FIG6_BINS, density=True, alpha=0.5,
        color='C0', label='True $\\eta$')
ax.hist(det_1g['eta_meas'], bins=FIG6_BINS, density=True, alpha=0.5,
        color='C3', label='Measured $\\eta$')
ax.plot(eta_th, arcsine_pdf, 'k--', lw=1.2, alpha=0.6, label='Arcsine')
ax.set_xlim(0, 1)
ax.set_ylim(0, FIG6_YLIM_HIST)
ax.set_xlabel('Splitting ratio $\\eta$')
ax.set_ylabel('Probability density')
ax.legend(fontsize=14)

# (b) η distribution — 10 GHz + SLD
ax = axes6[0, 1]
ax.hist(det_10g['eta_true'], bins=FIG6_BINS, density=True, alpha=0.5,
        color='C0', label='True $\\eta$')
ax.hist(det_10g['eta_meas'], bins=FIG6_BINS, density=True, alpha=0.5,
        color='C3', label='Measured $\\eta$')
ax.plot(eta_th, arcsine_pdf, 'k--', lw=1.2, alpha=0.6, label='Arcsine')
ax.set_xlim(0, 1)
ax.set_ylim(0, FIG6_YLIM_HIST)
ax.set_xlabel('Splitting ratio $\\eta$')
ax.set_ylabel('Probability density')
ax.legend(fontsize=14)

# (c) η autocorrelation
ax = axes6[1, 0]
ci99_det = float(det_10g['ci99'])
lags_d = np.arange(min(len(det_1g['ac_true']), FIG6_MAX_LAG))
ax.plot(lags_d[1:], det_1g['ac_meas'][1:len(lags_d)], 's-', ms=3, lw=1.0,
        color='C2', alpha=0.8, label='1 GHz free (meas.)')
ax.plot(lags_d[1:], det_10g['ac_true'][1:len(lags_d)], 'o-', ms=3, lw=1.0,
        color='C0', alpha=0.8, label='10 GHz+SLD (true)')
ax.plot(lags_d[1:], det_10g['ac_meas'][1:len(lags_d)], '^-', ms=3, lw=1.0,
        color='C3', alpha=0.8, label='10 GHz+SLD (meas.)')
ax.axhline(+ci99_det, color='r', ls='--', lw=1,
           label=f'99% CI = ±{ci99_det:.5f}')
ax.axhline(-ci99_det, color='r', ls='--', lw=1)
ax.axhline(0, color='gray', ls='-', lw=0.5)
ax.set_xlabel('Lag $k$')
ax.set_ylabel('$\\eta$ autocorrelation $C_\\eta(k)$')
ax.legend(fontsize=14, loc='upper right')
ax.set_xlim(0, FIG6_MAX_LAG)

# (d) ENOB sensitivity
ax = axes6[1, 1]
ax.plot(det_enob['enob'], det_enob['ac1_vs_enob'], 'o-', ms=5, lw=1.5,
        color='C0')
ax.axhline(+ci99_det, color='r', ls='--', lw=1,
           label=f'99% CI = ±{ci99_det:.5f}')
ax.axhline(-ci99_det, color='r', ls='--', lw=1)
ax.axhline(0, color='gray', ls='-', lw=0.5)
ax.axvline(det_p['scope_enob'], color='green', ls=':', lw=1.5,
           label=f'Nominal ({det_p["scope_enob"]:.1f} bits)')
ax.set_xlabel('ENOB (bits)')
ax.set_ylabel('Lag-1 $\\eta$ autocorrelation $C_\\eta(1)$')
ax.legend(fontsize=14)

plt.tight_layout()
if FIG6_SAVE:
    fig6.savefig(OUT / 'detection_chain.png',
                 dpi=FIG6_DPI, bbox_inches='tight')
    print(f'Saved: {OUT}/detection_chain.png')
plt.show()
```

```python
import numpy as np
import json
from pathlib import Path

DATA = Path('../images/acceptance_bw_comparison_v2/data')
bw_list = [0.6, 3.0, 8.0]
freq_labels = ['1 GHz', '2 GHz', '5 GHz', '8 GHz', '10 GHz']

sld = {bw: np.load(DATA / f'sld_bw{bw:.1f}nm.npz', allow_pickle=True) for bw in bw_list}
scalars = {bw: json.loads((DATA / f'scalars_bw{bw:.1f}nm.json').read_text()) for bw in bw_list}
```

## Figures 5.8–5.10 — Phase Transition Analysis

Data from `images/phase_transition/phase_transition_data_9freq.npz` (1M pulses, λ=1547 nm, L=150 µm, sinusoidal drive).

Frequencies: [1, 2, 3, 4, 5, 6, 7, 8, 10] GHz × 24 injection levels. Frequencies 1–2 GHz have no coherence transition (already fully randomised).

```python
# ── Load phase transition data ──────────────────────────────────
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

PT_DATA = Path('../images/phase_transition')
PT_OUT  = PT_DATA

d = np.load(PT_DATA / 'phase_transition_data_9freq.npz', allow_pickle=True)

target_freqs = d['target_freqs']          # [1e9, 2e9, ..., 1e10]
S_inj_values = d['S_inj_values']          # 24 log-spaced values
R1_CRIT      = float(d['R1_CRIT'])        # 1/e
DUTY         = float(d['DUTY'])
N_PT         = int(d['N_PULSES'])

# Fit parameters
beta_fit = float(d['beta_fit'])
beta_err = float(d['beta_err'])
C_fit    = float(d['C_fit'])
a_fit    = float(d['a_fit'])
b_fit    = float(d['b_fit'])
kappa_crit = float(d['kappa_crit'])

def freq_key(f):
    """Convert frequency (Hz) to npz dict key suffix matching _clean()."""
    return repr(float(f)).replace('.', 'p').replace('-', 'm').replace('+', '')

# Per-frequency S_c and colours
S_c = {}
freq_colors_pt = {}
for f in target_freqs:
    fk = freq_key(f)
    S_c[f] = float(d[f'S_c.{fk}'])
    freq_colors_pt[f] = str(d[f'freq_colors.{fk}'])

# Which frequencies had a real transition (S_c < max sweep)
S_inj_max = S_inj_values[-1]
fit_freqs = [f for f in target_freqs if S_c[f] < 0.99 * S_inj_max]
skip_freqs = [f for f in target_freqs if f not in fit_freqs]
has_transition = np.array([f in fit_freqs for f in target_freqs])

# Per-frequency, per-injection-level results
n_sinj = len(S_inj_values)
results_pt = {}
for f in target_freqs:
    fk = freq_key(f)
    res_list = []
    for i in range(n_sinj):
        res_list.append({
            'S_inj': float(d[f'results.{fk}.{i}.S_inj']),
            'r1':    float(d[f'results.{fk}.{i}.r1']),
        })
    results_pt[f] = res_list

# Theory curves (old standard phasor, kept for reference)
kappa_theory = d['kappa_theory']
r1_theory    = d['r1_theory']

# Arrays for all frequencies
f_arr     = np.array([f * 1e-9 for f in target_freqs])
Sc_arr    = np.array([S_c[f] for f in target_freqs])

# T_off/tau_p ratios stored in npz; recover tau_p from them
ratio_arr = np.array(d['ratio_arr'])          # all 9 freqs
tau_p = (1 - DUTY) / (target_freqs[0] * ratio_arr[0])

# Fit-frequency arrays for S_c scaling plot
fit_f_arr = np.array([f * 1e-9 for f in fit_freqs])
fit_Sc_arr = np.array([S_c[f] for f in fit_freqs])
fit_ratio_arr = ratio_arr[has_transition]

# Fit line for S_c scaling
ratio_fit = np.linspace(fit_ratio_arr.min() * 0.8, fit_ratio_arr.max() * 1.2, 100)
Sc_fit_line = np.exp(a_fit + b_fit * ratio_fit)

# Critical line for phase diagram (only over fit range)
f_dense = np.linspace(fit_f_arr.min(), fit_f_arr.max(), 100)
ratio_dense = (1 - DUTY) / (f_dense * 1e9 * tau_p)
Sc_dense = np.exp(a_fit + b_fit * ratio_dense)

tau_eff_ps = -1 / b_fit * tau_p * 1e12
tau_eff_ratio = -1 / b_fit

print(f'Phase transition data loaded: {len(target_freqs)} frequencies, '
      f'{n_sinj} injection levels, {N_PT//1000}k pulses')
print(f'tau_p = {tau_p*1e12:.2f} ps')
print(f'Transition at: {", ".join(f"{f*1e-9:.0f} GHz" for f in fit_freqs)}')
print(f'No transition: {", ".join(f"{f*1e-9:.0f} GHz" for f in skip_freqs)}')
print(f'S_c = ' + ', '.join(f'{f*1e-9:.0f}: {S_c[f]:.2e}' for f in fit_freqs))
print(f'beta = {beta_fit:.3f} ± {beta_err:.3f}')
print(f'T_off/tau_p range: {ratio_arr[-1]:.1f} (10 GHz) to {ratio_arr[0]:.1f} (1 GHz)')
print(f'S_c fit: exp({b_fit:.4f} * T_off/tau_p), tau_eff = {tau_eff_ps:.1f} ps ({tau_eff_ratio:.1f}x tau_p)')
```

```python
# ── Gain-enhanced random-phasor model ──────────────────────────
from scipy.interpolate import interp1d
from scipy.optimize import brentq, minimize_scalar

def _compute_phasor_r1(A_values, n=500_000, seed=42):
    """r1 from E = A + Z, Z ~ CN(0,1), via Monte Carlo."""
    rng = np.random.default_rng(seed)
    Z = (rng.standard_normal(n) + 1j * rng.standard_normal(n)) / np.sqrt(2)
    return np.array([np.abs(np.mean(np.exp(1j * np.angle(A + Z))))
                     for A in A_values])

# Build lookup table: r1 as a function of coherent amplitude A
_A_grid = np.logspace(-4, 3.5, 400)
_r1_grid = _compute_phasor_r1(_A_grid)
_r1_of_logA = interp1d(np.log10(_A_grid), _r1_grid,
                        kind='cubic', bounds_error=False,
                        fill_value=(_r1_grid[0], _r1_grid[-1]))

# Critical amplitude: A_c where r1(A_c) = 1/e
_logA_c = brentq(lambda la: float(_r1_of_logA(la)) - 1.0/np.e, -4, 3.5)
A_c = 10**_logA_c

def phasor_model(kappa, g=1.0):
    """Gain-enhanced phasor model: A_eff = A_c * kappa^{-g/2}.
    Normalised so r1 = 1/e at kappa = 1 for all g.
    """
    log_A = np.log10(A_c) - (g / 2) * np.log10(np.clip(kappa, 1e-10, None))
    return _r1_of_logA(log_A)

# Collect all (kappa, r1) data from frequencies with a transition
_all_kappa = []
_all_r1 = []
for _f in fit_freqs:
    _sc = S_c[_f]
    for _r in results_pt[_f]:
        _all_kappa.append(_r['S_inj'] / _sc)
        _all_r1.append(_r['r1'])
_all_kappa = np.array(_all_kappa)
_all_r1 = np.array(_all_r1)

# Fit g by minimising sum of squared residuals over all data
_sse = lambda g: np.sum((_all_r1 - phasor_model(_all_kappa, g=g))**2)
g_fit = minimize_scalar(_sse, bounds=(0.5, 5.0), method='bounded').x

# Theory curves
kp_model = np.logspace(-2.5, 3.5, 600)
g_fit = 3.34
r1_standard_model = phasor_model(kp_model, g=1.0)
r1_fitted_model   = phasor_model(kp_model, g=g_fit)

print(f'A_c = {A_c:.4f}  (r1 = {float(_r1_of_logA(_logA_c)):.6f} = 1/e)')
print(f'Fitted gain exponent g = {g_fit:.2f}')
print(f'Asymptotic tail exponent 2*beta = {2*beta_fit:.2f}')
print(f'At kappa=1: std r1 = {phasor_model(np.array([1.0]), 1.0)[0]:.4f}, '
      f'fit r1 = {phasor_model(np.array([1.0]), g_fit)[0]:.4f} (both 1/e)')
print(f'At kappa=5: std r1 = {phasor_model(np.array([5.0]), 1.0)[0]:.4f}, '
      f'fit r1 = {phasor_model(np.array([5.0]), g_fit)[0]:.4f}')
```

```python
# ─── Fig 5.8: Universal data collapse — Tweakable ─────────────
FIG8_FIGSIZE   = (16, 6)
FIG8_MS        = 5
FIG8_XLIM_LIN  = (0, 15)
FIG8_DPI       = 150
FIG8_SAVE      = True
# ───────────────────────────────────────────────────────────────

fig8, axes8 = plt.subplots(1, 2, figsize=FIG8_FIGSIZE)

# Noise floor: 99% CI for r1 under null (uniform iid phases)
r1_noise_floor = np.sqrt(-np.log(0.01) / N_PT)

# (a) Linear scale
ax = axes8[0]
for f_rep in target_freqs[2:]:
    fg = f_rep * 1e-9
    res = results_pt[f_rep]
    s_c = S_c[f_rep]
    kappa = [r['S_inj'] / s_c for r in res]
    r1 = [r['r1'] for r in res]
    col = freq_colors_pt[f_rep]
    ax.plot(kappa, r1, 'o', color=col, ms=FIG8_MS, alpha=0.7,
            label=f'{fg:.0f} GHz')

ax.plot(kp_model, r1_fitted_model, 'k-', lw=2.5, alpha=0.7,
        label=f'Phasor model ($g = {g_fit:.2f}$)', zorder=0)
ax.axhline(R1_CRIT, color='gray', ls='--', lw=1, alpha=0.3)
ax.axvline(1.0, color='gray', ls='--', lw=1, alpha=0.3)
ax.set_xlabel('$\\kappa = S_{inj} / S_c$')
ax.set_ylabel('$r_1$')
ax.set_xlim(*FIG8_XLIM_LIN)
ax.set_ylim(-0.02, 1.05)
ax.legend(fontsize=11, ncol=2)

# (b) Log-log scale — exclude noise-floor points (< 10x CI)
ax = axes8[1]
for f_rep in target_freqs[2:]:
    fg = f_rep * 1e-9
    res = results_pt[f_rep]
    s_c = S_c[f_rep]
    kappa = np.array([r['S_inj'] / s_c for r in res])
    r1 = np.array([r['r1'] for r in res])
    mask = (kappa > 0.05) & (r1 > 10 * r1_noise_floor)
    col = freq_colors_pt[f_rep]
    ax.loglog(kappa[mask], r1[mask], 'o', color=col, ms=FIG8_MS, alpha=0.7,
              label=f'{fg:.0f} GHz')

mask_f = (kp_model > 0.05) & (r1_fitted_model > 0.001)
ax.loglog(kp_model[mask_f], r1_fitted_model[mask_f], 'k-', lw=2.5,
          alpha=0.7, label=f'Phasor model ($g = {g_fit:.2f}$)', zorder=0)
kappa_ref = np.logspace(0.5, 3, 50)
ax.loglog(kappa_ref, C_fit * kappa_ref**(-beta_fit), 'r--', lw=1.5,
          label=f'$r_1 \\propto \\kappa^{{-{beta_fit:.2f}}}$')
ax.axhline(R1_CRIT, color='gray', ls='--', lw=0.8, alpha=0.3)
ax.axvline(1.0, color='gray', ls='--', lw=0.8, alpha=0.3)
ax.set_xlabel('$\\kappa = S_{inj} / S_c$')
ax.set_ylabel('$r_1$')
ax.legend(fontsize=11)

plt.tight_layout()
if FIG8_SAVE:
    fig8.savefig(PT_OUT / 'universal_collapse.png', dpi=FIG8_DPI, bbox_inches='tight')
    print(f'Saved: {PT_OUT}/universal_collapse.png')
plt.show()
```

## Figures — Timing Jitter Analysis

Data from `images/timing_jitter/timing_jitter_data.npz` (1M pulses, sinusoidal drive, 1–10 GHz).

Figures:
- **Jitter vs frequency** — absolute, period, and phase jitter (free vs SLD)
- **Arrival distributions** — peak arrival time histograms at selected frequencies
- **Jitter trade-off** — timing vs phase jitter parametric plot + SLD ratio
- **Correlations at 10 GHz** — *requires raw per-pulse data; run script directly*
- **Allan deviation** — timing stability at selected frequencies

```python
# ── Load timing jitter data ─────────────────────────────────────
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

TJ_DATA = Path('../images/timing_jitter')
TJ_OUT  = TJ_DATA

tj = np.load(TJ_DATA / 'timing_jitter_data.npz', allow_pickle=True)

tj_freqs = tj['freqs']            # [1e9, 2e9, ..., 1e10]
tj_f_ghz = tj['f_ghz']            # [1, 2, ..., 10]
tj_N     = int(tj['N_PULSES'])
tj_DUTY  = float(tj['DUTY'])
tj_I_DC  = float(tj['I_DC'])
tj_V_RF  = float(tj['V_RF_AMP'])
tj_S_INJ = float(tj['S_INJ'])

# Unpack per-frequency, per-case scalar metrics into convenient dicts
tj_data = {}
for fg in tj_f_ghz:
    for case in ['free', 'sld']:
        key = f'{fg:.0f}GHz_{case}'
        tj_data[key] = {
            'sig_t':   float(tj[f'data.{key}.sig_t']),
            'sig_T':   float(tj[f'data.{key}.sig_T']),
            'sig_phi': float(tj[f'data.{key}.sig_phi']),
            'r1':      float(tj[f'data.{key}.r1']),
            'f_rep':   float(tj[f'data.{key}.f_rep']),
            'T_rep':   float(tj[f'data.{key}.T_rep']),
        }
        # Arrival time histogram (t_peak was > 200k so stored as histogram)
        tj_data[key]['t_peak_counts'] = tj[f'data.{key}.t_peak__hist_counts']
        tj_data[key]['t_peak_edges']  = tj[f'data.{key}.t_peak__hist_edges']
        tj_data[key]['t_peak_stats']  = tj[f'data.{key}.t_peak__stats']
        # Allan deviation (small arrays, stored directly)
        tj_data[key]['tau_ad'] = tj[f'data.{key}.tau_ad']
        tj_data[key]['adev']   = tj[f'data.{key}.adev']

print(f'Timing jitter data loaded: {len(tj_f_ghz)} frequencies, '
      f'{tj_N//1000}k pulses')
print(f'I_DC={tj_I_DC*1e3:.1f} mA, V_RF={tj_V_RF} V, '
      f'S_inj={tj_S_INJ:.2e} m^-3, duty={tj_DUTY:.2f}')
print(f'Jitter @ 10 GHz: free={tj_data["10GHz_free"]["sig_t"]*1e12:.2f} ps, '
      f'sld={tj_data["10GHz_sld"]["sig_t"]*1e12:.2f} ps')
```

```python
# ─── Arrival distributions — Tweakable ─────────────────────────
TJ2_SEL_GHZ = [1, 3, 5, 10]   # frequencies to overlay
TJ2_FIGSIZE = (16, 6)
TJ2_DPI     = 150
TJ2_SAVE    = True
TJ2_REBIN   = 8               # combine this many adjacent bins (512 → 64)
# ───────────────────────────────────────────────────────────────

fig_tj2, axes_tj2 = plt.subplots(1, 2, figsize=TJ2_FIGSIZE)
colors_tj = plt.cm.viridis(np.linspace(0.1, 0.9, len(TJ2_SEL_GHZ)))

for ax, (case, title) in zip(axes_tj2, [
    ('free', 'Free-running'), ('sld', 'SLD-injected')
]):
    for i, fg in enumerate(TJ2_SEL_GHZ):
        key = f"{fg}GHz_{case}"
        d = tj_data[key]
        counts = d['t_peak_counts']
        edges  = d['t_peak_edges']
        stats  = d['t_peak_stats']

        # Rebin
        n = TJ2_REBIN
        n_new = len(counts) // n
        counts_rb = counts[:n_new * n].reshape(n_new, n).sum(axis=1)
        edges_rb = edges[::n][:n_new + 1]

        widths = np.diff(edges_rb)
        density = counts_rb / (counts_rb.sum() * widths)

        edges_ps = edges_rb * 1e12
        ax.stairs(density, edges_ps, color=colors_tj[i], lw=1.5,
                  label=f'{fg} GHz ($\\sigma$ = {stats[1]*1e12:.1f} ps)')

    ax.set_xlabel('Peak arrival time (ps)', fontsize=14)
    ax.set_ylabel('Probability density', fontsize=14)
    # ax.set_title(title)
    ax.legend(fontsize=14)

plt.tight_layout()
if TJ2_SAVE:
    fig_tj2.savefig(TJ_OUT / 'arrival_distributions.png', dpi=TJ2_DPI, bbox_inches='tight')
    print(f'Saved: {TJ_OUT}/arrival_distributions.png')
plt.show()
```

## Figure — Injection Requirements (S_inj Sweep)

Data from `images/qkd_source/qkd_sinj_sweep_data.npz` (500k pulses, 25 S_inj values, 3 frequencies: 3, 5, 10 GHz).

Figures:
- **r1 vs S_inj** — phase correlation against injected photon density
- **Jitter cost** — timing jitter, phase jitter, intensity CV against S_inj
- **Security vs performance** — parametric trade-off (r1 vs jitter/CV)

```python
# ── Load S_inj sweep data ──────────────────────────────────────
SINJ_DATA = Path('../images/qkd_source')
SINJ_OUT  = SINJ_DATA

sinj = np.load(SINJ_DATA / 'qkd_sinj_sweep_data.npz', allow_pickle=True)

sinj_S_values = sinj['S_inj_values']
sinj_freqs = sinj['target_freqs']
R1_THRESHOLD = float(sinj['R1_THRESHOLD'])

def sinj_freq_key(f):
    return repr(float(f)).replace('.', 'p').replace('-', 'm').replace('+', '')

sinj_colors = {}
sinj_baselines = {}
sinj_results = {}
sinj_thresholds = {}

for f_rep in sinj_freqs:
    fk = sinj_freq_key(f_rep)
    sinj_colors[f_rep] = str(sinj[f'colors.{fk}'])

    sinj_baselines[f_rep] = {
        'r1':      float(sinj[f'baselines.{fk}.r1']),
        'sig_t':   float(sinj[f'baselines.{fk}.sig_t']),
        'sig_phi': float(sinj[f'baselines.{fk}.sig_phi']),
        'cv':      float(sinj[f'baselines.{fk}.cv']),
    }

    res = []
    for i in range(len(sinj_S_values)):
        res.append({
            'S_inj':   float(sinj[f'results.{fk}.{i}.S_inj']),
            'r1':      float(sinj[f'results.{fk}.{i}.r1']),
            'sig_t':   float(sinj[f'results.{fk}.{i}.sig_t']),
            'sig_phi': float(sinj[f'results.{fk}.{i}.sig_phi']),
            'cv':      float(sinj[f'results.{fk}.{i}.cv']),
        })
    sinj_results[f_rep] = res

    thr_key = f'thresholds.{fk}.S_inj'
    if thr_key in sinj:
        sinj_thresholds[f_rep] = {
            'S_inj':   float(sinj[thr_key]),
            'idx':     int(sinj[f'thresholds.{fk}.idx']),
            'metrics': {
                'r1':      float(sinj[f'thresholds.{fk}.metrics.r1']),
                'sig_t':   float(sinj[f'thresholds.{fk}.metrics.sig_t']),
                'sig_phi': float(sinj[f'thresholds.{fk}.metrics.sig_phi']),
                'cv':      float(sinj[f'thresholds.{fk}.metrics.cv']),
            },
        }
    else:
        sinj_thresholds[f_rep] = None

# Reference S_inj annotations
sinj_ref_points = {}
for k in sinj.files:
    if k.startswith('ref_points.'):
        label = k.replace('ref_points.', '').replace('p', '.').replace('_', ' ')
        sinj_ref_points[label] = float(sinj[k])

print(f'S_inj sweep loaded: {len(sinj_freqs)} frequencies, '
      f'{len(sinj_S_values)} injection levels')
for f_rep in sinj_freqs:
    fg = f_rep * 1e-9
    t = sinj_thresholds[f_rep]
    bl = sinj_baselines[f_rep]
    if t is not None:
        m = sinj_results[f_rep][t['idx']]
        print(f'  {fg:.0f} GHz: threshold S_inj={t["S_inj"]:.2e}, '
              f'jitter {bl["sig_t"]*1e12:.1f} -> {m["sig_t"]*1e12:.1f} ps '
              f'({m["sig_t"]/bl["sig_t"]:.1f}x), CV={m["cv"]*100:.0f}%')
    else:
        print(f'  {fg:.0f} GHz: threshold NOT reached')
```

## Figure — Scalability: Full Model vs Simplified Photon-Number Model

Compares the full rate-equation simulation (DUTY=0.30, τ_eff=24.9 ps from S_c scaling) with the
simplified photon-number model from the thesis scalability section (DUTY=0.50, τ_p=6.4 ps, R≥10).

- **Panel (a)**: Required SLD power vs repetition rate. Full model calibrated to the experimental
  point (19 mW at 10 GHz). Simplified model uses its native parameters from the thesis.
- **Panel (b)**: Maximum achievable repetition rate vs SLD power for different coupling improvements.
  Modulation bandwidth limit (18 GHz) shown as hard ceiling.

```python
# ─── Scalability comparison — Tweakable ────────────────────────
SCAL_FIGSIZE = (16, 6.5)
SCAL_DPI     = 150
SCAL_LW      = 2.0
SCAL_SAVE    = True
# ───────────────────────────────────────────────────────────────

from scipy.interpolate import interp1d

h_c  = 6.626e-34
c_l  = 3e8
nu0  = c_l / 1550e-9

# ═══ FULL SIMULATION MODEL (DUTY=0.30, L=150 µm DFB) ═══
tau_p_sim   = 2.07e-12   # bare cavity photon lifetime
tau_eff_sim = 24.9e-12   # effective off-state lifetime from S_c scaling
DUTY_sim    = 0.30

# Interpolated r1=0.01 thresholds from sinj sweep
sim_freqs    = np.array([3e9, 5e9, 10e9])
sim_S_interp = np.array([2.19e19, 5.827e19, 4.244e21])

# Calibrate coupling: 19 mW → S_inj=4.244e21 at 10 GHz
P_exp  = 19e-3
S_10   = sim_S_interp[2]
K_cal  = S_10 / P_exp

# Extrapolation using tau_eff, anchored at 10 GHz
T_off_10  = (1 - DUTY_sim) / 10e9
f_ext     = np.logspace(np.log10(1e9), np.log10(50e9), 300)
T_off_ext = (1 - DUTY_sim) / f_ext
S_thresh_full = S_10 * np.exp((T_off_10 - T_off_ext) / tau_eff_sim)
P_full_W = S_thresh_full / K_cal

# ═══ SIMPLIFIED MODEL (thesis §5.4.5, DUTY=0.50) ═══
tau_p_s   = 6.4e-12
DUTY_s    = 0.50
V_cm3_s   = 2e-11
Gamma_s   = 0.22
beta_s    = 4.44e-5
N0_s      = 3.79e16 * (9.0 / 14.7)
tau_n_s   = 0.15e-9
eta_d_s   = 0.52
eta_c_s   = 0.75
P_out_s   = 10e-3
Dv_sled_s = 4.1e12

E_gamma_s  = h_c * c_l / 1547e-9
Dv_cav_s   = 1 / (2 * np.pi * tau_p_s)
eta_spec_s = Dv_cav_s / Dv_sled_s
eta_pol_s  = 0.5
eta_s      = eta_spec_s * eta_pol_s * eta_c_s
Laser_N_s  = P_out_s * (2 * Gamma_s * tau_p_s) / (eta_d_s * E_gamma_s * eta_c_s)

T_off_s      = (1 - DUTY_s) / f_ext
N_residual_s = Laser_N_s * np.exp(-T_off_s / tau_p_s)
R_sp_s       = V_cm3_s * Gamma_s * beta_s * N0_s * np.exp(-T_off_s / tau_n_s) / tau_n_s
N_spont_s    = R_sp_s * tau_p_s
P_simple_W   = np.maximum((10 * N_residual_s - N_spont_s) * E_gamma_s / (eta_s * tau_p_s), 0)

# ═══ FIGURE ═══
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=SCAL_FIGSIZE, dpi=SCAL_DPI)
f_GHz = f_ext / 1e9

cf_styles = [(1, '-',  1.0, 'Baseline coupling'),
             (2, '--', 0.7, r'$2\times$ coupling'),
             (5, ':',  0.5, r'$5\times$ coupling')]

# ── Panel (a): P_SLD vs repetition rate ──
for cf, ls, alpha, lab in cf_styles:
    ax1.semilogy(f_GHz, P_full_W / cf * 1e3, color='C0', ls=ls, lw=SCAL_LW,
                 alpha=alpha, label=f'Full model, {lab}')

mask_s = P_simple_W > 0
ax1.semilogy(f_GHz[mask_s], P_simple_W[mask_s] * 1e3, 'k--', lw=SCAL_LW,
             label=f'Simplified ($D$={DUTY_s}, $\\tau_p$={tau_p_s*1e12:.1f} ps)')

P_sim_mW = sim_S_interp / K_cal * 1e3
ax1.scatter(sim_freqs / 1e9, P_sim_mW, marker='o', s=70, zorder=5,
            color='C0', edgecolors='k', linewidths=0.8,
            label='Simulation ($r_1 < 0.01$)')

ax1.scatter([10], [P_exp * 1e3], marker='*', s=250, color='gold',
            edgecolors='k', linewidths=0.8, zorder=6,
            label='Experiment (19 mW)')

for plim, lab in [(50, '50 mW'), (100, '100 mW'), (300, '300 mW')]:
    ax1.axhline(plim, color='gray', ls=':', alpha=0.4, lw=0.8)
    ax1.text(1.15, plim * 1.15, lab, fontsize=8, color='0.45', va='bottom')

ax1.axvspan(18, 55, color='0.88', alpha=0.4, zorder=0)
ax1.text(30, 3e3, 'Beyond modulation\nbandwidth', fontsize=9, color='0.4',
         ha='center', va='top', style='italic')

ax1.set_xlabel('Repetition rate (GHz)', fontsize=12)
ax1.set_ylabel('Required SLD power (mW)', fontsize=12)
ax1.set_xlim(1, 50)
ax1.set_ylim(0.01, 5e3)
ax1.legend(fontsize=8, loc='upper left', framealpha=0.9)
ax1.tick_params(axis='both', which='major', direction='in', length=5, width=1, labelsize=11)
ax1.tick_params(axis='both', which='minor', direction='in', length=3, width=0.7)
ax1.set_title('(a)  Required SLD power', fontsize=12, pad=8)

txt = (f'Full: $\\tau_{{\\mathrm{{eff}}}}$ = {tau_eff_sim*1e12:.0f} ps '
       f'({tau_eff_sim/tau_p_sim:.0f}× $\\tau_p$)\n'
       f'Simplified: $\\tau_{{p,\\mathrm{{eff}}}}$ = {tau_p_s*1e12:.1f} ps '
       f'({tau_p_s/tau_p_sim:.1f}× $\\tau_p$)')
ax1.text(0.98, 0.04, txt, transform=ax1.transAxes, fontsize=8.5,
         ha='right', va='bottom',
         bbox=dict(boxstyle='round,pad=0.4', fc='white', ec='0.7', alpha=0.9))

# ── Panel (b): max f_rep vs P_SLD ──
P_sld_range = np.logspace(np.log10(0.5e-3), np.log10(500e-3), 300)

def max_frep_full(P_sld, coupling_mult=1):
    P_eff = P_sld * coupling_mult
    ratio = P_eff / P_exp
    if ratio <= 0:
        return np.nan
    denom = T_off_10 - tau_eff_sim * np.log(ratio)
    if denom <= 0:
        return 50e9
    return min((1 - DUTY_sim) / denom, 50e9)

def max_frep_simple(P_sld_W):
    mask = P_simple_W <= P_sld_W
    if not np.any(mask):
        return f_ext[0]
    return min(f_ext[mask][-1], 50e9)

for cf, ls, alpha, lab in cf_styles:
    f_max = np.array([max_frep_full(p, cf) for p in P_sld_range])
    valid = np.isfinite(f_max) & (f_max > 1e9)
    ax2.semilogx(P_sld_range[valid] * 1e3, f_max[valid] / 1e9,
                 color='C0', ls=ls, lw=SCAL_LW, alpha=alpha,
                 label=f'Full model, {lab}')

f_max_s = np.array([max_frep_simple(p) for p in P_sld_range])
valid_s = f_max_s > 1e9
ax2.semilogx(P_sld_range[valid_s] * 1e3, f_max_s[valid_s] / 1e9, 'k--', lw=SCAL_LW,
             label=f'Simplified ($D$={DUTY_s})')

ax2.axhspan(18, 52, color='0.88', alpha=0.4, zorder=0)
ax2.text(200, 35, 'Beyond modulation\nbandwidth', fontsize=9, color='0.4',
         ha='center', va='center', style='italic')
ax2.axhline(18, color='C3', ls='-', lw=1.2, alpha=0.7)

ax2.scatter([P_exp * 1e3], [10], marker='*', s=250, color='gold',
            edgecolors='k', linewidths=0.8, zorder=6)

for pref, lab in [(6.7, '6.7 mW\n(this SLD)'), (100, '100 mW'), (300, '300 mW')]:
    ax2.axvline(pref, color='0.82', ls=':', lw=0.8, zorder=0)
    ax2.text(pref, 1.5, lab, fontsize=7, color='0.5', ha='center', va='bottom')

ax2.set_xlabel('Available SLD power (mW)', fontsize=12)
ax2.set_ylabel('Maximum repetition rate (GHz)', fontsize=12)
ax2.set_xlim(0.5, 500)
ax2.set_ylim(1, 50)
ax2.legend(fontsize=8, loc='lower right', framealpha=0.9,
           bbox_to_anchor=(0.98, 0.02))
ax2.tick_params(axis='both', which='major', direction='in', length=5, width=1, labelsize=11)
ax2.tick_params(axis='both', which='minor', direction='in', length=3, width=0.7)
ax2.set_title('(b)  Maximum achievable repetition rate', fontsize=12, pad=8)

fig.tight_layout()

if SCAL_SAVE:
    fig.savefig(SINJ_OUT / 'scalability_comparison.png', dpi=300,
                bbox_inches='tight', pad_inches=0.1)
    print('Saved: scalability_comparison.png')

plt.show()

print('\n── Key predictions (full model, baseline coupling) ──')
for f_ghz in [10, 15, 20, 25, 30]:
    p = S_10 * np.exp((T_off_10 - (1-DUTY_sim)/(f_ghz*1e9))/tau_eff_sim) / K_cal * 1e3
    print(f'  {f_ghz} GHz: {p:.1f} mW')
print(f'\n── Simplified model ──')
for f_ghz in [10, 15, 20]:
    idx = np.argmin(np.abs(f_ext - f_ghz*1e9))
    print(f'  {f_ghz} GHz: {P_simple_W[idx]*1e3:.1f} mW')
print(f'\n── Max rate at given SLD power (full model) ──')
for p_mw in [19, 50, 100, 300]:
    fm = max_frep_full(p_mw*1e-3, 1)
    print(f'  {p_mw} mW → {min(fm/1e9, 18):.1f} GHz (injection limit: {fm/1e9:.1f} GHz)')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import ScalarFormatter

# Code to simulate SLED injection into a laser diode 
# Based on tracking the evolution of the photon number directly after the laser is gated OFF.
# The laser photon number decays exponentially.
# The spontaneous emission photon number is modulated by the decaying carrier concentration (also an exponential decay)
# The SLED input is assumed to be continuous and via the photon lifetime, achieves a steady state in the laser cavity.
# We then compute the ratio of the incoherent light to the coherent light (SLED + spont / laser) to work out if we are in the phase randomised
# regime, ie. SLED + spont / laser > 1

# --- Physical constants ---
h = 6.626e-34 # planck's constant
c = 299792458 # speed of light
lam = 1547e-9 # wavelength of laser diode used in the experiment
E_gamma = h * c / lam

# --- System parameters ---
# --- Note - almost all parameters were taken directly from Victor's paper: https://journals.aps.org/prapplied/pdf/10.1103/PhysRevApplied.16.054012 ---
tau_p = 6.4e-12 # photon lifetime. in Victor's paper this is set to 4.47ps. However I have tuned this so it is a little bit longer and this seems to 
                # give curves that better resemble the experiment. It's entirely possible the residual gain at and slightly below threshold reduces 
                # the net decay rate, leading to an effective photon lifetime longer than the bare cavity lifetime.
                # best value seeems to be 6.4ps

# --- Laser device parameters (from paper) ---
tau_n = 0.15e-9     # carrier lifetime (important for spontaneous emission through e-h recombination)
V = 2e-11           # volume of the cavity (cm^-3)
N0 = 3.79e16 * ( 9.0 / 14.7 )       # carrier density at transparency (cm^-3); assume this is the density at threshold and can be calculated from the threshold curent. 
                                    # in the case fromm the paper, the threshold current was 14.7mA. Adam measured 9mA so we scale accordingly. 
beta = 4.44e-5      # fraction of spontaneous emission coupled into the lasing mode
Gamma = 0.22        # confinement factor
eta_d = 0.52        # differential quantum efficiency
P_out = 10e-3       # laser optical output - assume 10mW for now

# --- SLED bandwidth ---
Dv_sled = 4.1e12    # assumes spread of 33nm at 1550nm. 

# --- SLED spectral ovelap with laser cavity ---
Dv_cav = 1 / (2 * np.pi * tau_p) # cavity bandwidth from SLED paper eqn. 4. 

# SLED coupling model completely separate
eta_spec = Dv_cav / Dv_sled                             # accounts for the spectral overlap of the laser mode and the SLD broad spectrum
eta_pol = 0.5                                           # accounts for the fact the SLED is unpolarised but laser cavity is polarised
eta_coupling = 0.75                                     # coupling of laser to facet and fibre, assume 75%
eta_spatial = 1.0                                       # accounts for the spatial overlap of the incoming mode with the internal laser cavity mode
eta = eta_spec * eta_pol * eta_spatial * eta_coupling   # SLED coupling efficiency

# Updated TOTAL photon number using paper-consistent mapping (eqn. 7 in Victor's paper). Remember to remove volume term since we want total photon number not photon number density
Laser_photon_number = P_out * (2 * Gamma * tau_p) / (eta_d * E_gamma * eta_coupling )

# --- Power axis (fine + coarse) ---
P_mW_fine = np.arange(0, 1, 0.001)
P_mW_coarse = np.arange(1, 500, 1)
P_mW = np.concatenate((P_mW_fine, P_mW_coarse))
P = P_mW * 1e-3

# --- Time windows ---
tau_windows = [
    # 500e-12,    # 1 GHz
    100e-12,    # 5 GHz
    # 62.5e-12,   # 8 GHz
    50e-12,     # 10 GHz
    #33.33e-12,  # 15 GHz
    25e-12,     # 20 GHz
    #10e-12,     # 50 GHz
    # 5e-12       # 100 GHz
]

# =========================
# Plot: Ratio 
# =========================
plt.figure()

for tau_w in tau_windows:

    # Effective "clock rate" label for each OFF window
    f_window = 1 / (2 * tau_w)
    f_window_GHz = f_window * 1e-9

    # --- Laser spontaneous floor ---
    # Use the carrier density from when the laser is just switched off, ie at transparency N0 and plug into term 4 of eqn. 5 in Victor's paper. 
    # Then assume an exponential decay of the carrier density during the OFF period (ie. no current)
    R_sp =  V * Gamma * beta * N0  * np.exp( -tau_w / tau_n) / tau_n
    N_spont_laser = R_sp * tau_p

    # --- SLED photon number ---
    R_in = (P * eta) / E_gamma
    N_sled = R_in * (tau_p)

    # --- Laser photon number ---
    # Solution to qn. 5 in the Victor paper assuming first term is zero (no stim emission)
    N_residual = Laser_photon_number * np.exp(-tau_w / tau_p)
    N_laser = N_residual + N_spont_laser

    print(f"tau_w = {tau_w*1e12:.1f} ps")
    print(f"N_residual = {N_residual:.2e}")
    print(f"N_spont = {N_spont_laser:.2e}")
    print()

    # --- Ratio ---
    # Compute ratio of random light to laser (coherent) light
    N_residual_safe = max(N_residual, 1e-20) # avoid divide by zero blowup
    ratio = (N_sled + N_spont_laser) / N_residual_safe

    # Plot 
    plt.plot(P_mW, ratio, label=f"{f_window_GHz:.1f} GHz")

# Formatting
plt.xlabel("SLD Output Power (mW)", fontsize = 14)
plt.ylabel("R", fontsize = 14)
plt.axhline(10, linestyle="--", color="black", label="Randomisation threshold") # set this to 10, so well above the transition between coherent 
                                                                                # and phase randomisation
ax = plt.gca()
formatter = ScalarFormatter(useMathText=True)
formatter.set_scientific(True)
formatter.set_powerlimits((0, 0))
# ax.yaxis.set_major_formatter(formatter)
ax.yaxis.get_offset_text().set_fontsize(14)
ax.xaxis.get_offset_text().set_fontsize(14)
ax.tick_params(axis='both', which='major', direction='in', length=6, width=1.2, labelsize=14)
ax.tick_params(axis='both', which='minor', direction='in', length=3, width=1.0)
ax.minorticks_on()
# plt.title("Ratio of N_SLD + N_spont to N_laser")
plt.grid(True)
# plt.legend(title="Clock rate")
plt.legend(fontsize="14")
# Optional: log scale
plt.yscale("log")
plt.xscale("log")
plt.show()
```
